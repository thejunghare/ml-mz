```python
import numpy as np
```

```python
arr = np.array(1) # basic array in numpy
arr1 = np.array([1,2,3,4,5,6,7,8,8])
print(arr1)

# some maths
print(arr1 * 2)

# access elements
print(arr1[1]) # 2

# slice
print(arr1[1:4]) # 2,3,4
print(arr1[:4]) #  1,2,3,4
print(arr1[3:])
print(arr1[1:6:2]) #2, 4, 6
print(arr1[:-1]) #  1- 7
print(arr1[-1:]) # 8


arr2 = np.array([[1,2,3,4], [5,6,7,8]])
print(arr2[1,0])

print(arr2[1, : 3])
```

```python
# mean
arr1_mean = np.mean(arr1)
# median
arr1_median = np.median(arr1)
# mode
arr1_mode = np.unique(arr1)
# std
arr1_std = np.std(arr1)
print(arr1_mean, arr1_median, arr1_mode, arr1_std)
```
