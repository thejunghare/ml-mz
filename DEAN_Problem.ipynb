```python
import pandas as pd
```

```python
url = 'https://raw.githubusercontent.com/caesarw0/ml-dataset/main/students_dropout_prediction/data.csv'
pd.options.display.max_rows = 9999
df = pd.read_csv(url, sep=";")

print(df)
```

```python
df['dropped_out'] = df['Target'] == 'Dropout'
df.head(10)
df['dropped_out'] = df['dropped_out'].astype(int)
df.head(10)
```

```python
cr = df.corr(numeric_only=True)
cr_with_do = cr.abs()
print(cr_with_do['dropped_out'].sort_values(ascending=False))
```
