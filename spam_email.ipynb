```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
# null values
# fill null values
```

```python
df = pd.read_excel('/content/email_spam_dataset_rich.xlsx')
```

```python
# unq, null
print(df.isnull().sum())
```

```python
print(df.Label)
```

```python
# df_copy = df.copy()

df['Label'] = df['Label'] == 'Spam'
print(df.head())
```

```python
df['Label'] = df['Label'].astype(int)
print(df.head())
```

```python
x = df[['NumAttachments', 'NumLinks']]
X_train, X_test, y_train, y_test=train_test_split(x, df.Label, random_state=0)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn = KNeighborsClassifier(n_neighbors =2)
```

```python
knn.fit(X_train, y_train)
knn.predict(X_test)
```
